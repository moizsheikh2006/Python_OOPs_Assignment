**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
# Write your code here for Q1
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number 
        
    #getter
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    #setter
    def set_name (self,name):
        self.__name = name

    def set_roll_number (self,roll_number):
        self.__roll_number = roll_number



s1 = Student("Moiz", 100)

print (s1.get_name())
print(s1.get_roll_number())


s1.set_name("Ahmed")
s1.set_roll_number(101)


print(s1.get_name())
print(s1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
# Write your code here for Q2

class BankAcount:
    def __init__(self,balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance+=amount
            print(f"Rs. {amount} deposited succesfully!")
            print(f"Current balance: Rs. {self.__balance}")
        else:
            print(f"Invalid deposit amount: Rs. {amount}")

    def withdraw(self, amount):
        if amount <= 0:
            print (f"Invalid withdrawl amount: Rs. {amount}")
        elif amount <= self.__balance:
            self.__balance-= amount
            print (f"Rs. {amount} withdrawn succesfully!")
            print (f"Remaning balance Rs. {self.__balance}")
        else:
            print (f"Insufficient balance! Your balance is Rs. {self.__balance}")

    def get_balance(self):
        return self.__balance


account = BankAcount(50000)

account.deposit(20000)
account.withdraw(30000)
account.withdraw(50000)

print(f"Final balance: Rs. {account.get_balance()}")
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
# Write your code here for Q10

# Hierarhical Inheritance main ek parent class se 2 ya 2 se zyda child casses nikal sakti hain.
class Shape:
    def show(self):
        print("This is a shape.")

class Circle(Shape):
    def area (self, radius):
        print(f"Circle Area {3.14 * radius * radius}")

class Square(Shape):
    def area(self, side):
        print(f"Square Area: {side * side}")

circle = Circle()
circle.show()
circle.area(5)

square = Square()

square.show()
square.area(4)
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
# Write your code here for Q11

#Method Overriding tab hota hai jb child class parent class ke method ko apne tareeqe se dobara define kri hai. Method ka naam same hota hai lekin uska behavior change hota hai
class Bird: 
    def fly(self):
        print("Most birds can fly")

class Ostrich(Bird):
    def fly (self):
        print("Ostrich cannot fly")


bird = Bird()
bird.fly()


ostrich = Ostrich()
ostrich.fly()
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
# Write your code here for Q13

#Operator Overloading ka matlb hai kisi operator jaise +,- ya * ko apni class ke object ke liye custom behaviour dena.

class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

number1 = Number(10)
number2 = Number(30)

result = number1 + number2

print(f"Result: {result}")
```

**Q17:** Q16 wali `Computer` class ko inherit karke do child classes `Laptop` aur `Desktop` banayein aur dono mein `process()` method ko mandatory implement karein.

```python
# Write your code here for Q17

from abc import ABC, abstractmethod

class Computer(ABC):
    @abstractmethod
    def process(self):
        pass

class Laptop(Computer):
    def process(self):
        print("Laptop is processing data.")

class Desktop(Computer):
    def process(self):
        print ("Desktop is processing data.")


laptop = Laptop()
desktop = Desktop()

laptop.process()
desktop.process()
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
# Write your code here for Q18

from abc import ABC, abstractmethod

class Car(ABC):
    def start_engine(self):
        print("Engine started!")

        @abstractmethod
        def drive(self):
            pass

class SportsCar(Car):
    def drive(self):
        print ("Sports car is driving!")

sportscar = SportsCar()

sportscar.start_engine()
sportscar.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
# Write your code here for Q20

from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        print(f"Credit Card payment: Rs. {amount} successfull!")

class UPIPayment(Payment):
    def pay(self, amount):
        print(f"UPI payment: Rs. {amount} successfull!")

p1 = CreditCardPayment()
p2 = UPIPayment()

p1.pay(50000)
p2.pay(20000)
```
